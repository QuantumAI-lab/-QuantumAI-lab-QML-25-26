# One-Qubit Gates with Qiskit
This notebook builds a one‑qubit register, applies various gates, shows the statevector, and draws circuits.

```python
from math import pi
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from IPython.display import display
```

## Basic gates

```python
qc_x = QuantumCircuit(1)
qc_x.x(0)
display(qc_x.draw(output='mpl'))
sv_2 = Statevector.from_instruction(qc_x)
display(sv_2.draw('latex', prefix="|ψ⟩ = "))
```

```python
qc_y = QuantumCircuit(1)
qc_y.y(0)
display(qc_y.draw(output='mpl'))
sv_3 = Statevector.from_instruction(qc_y)
display(sv_3.draw('latex', prefix="|ψ⟩ = "))
```

```python
qc_z = QuantumCircuit(1)
qc_z.x(0)
qc_z.z(0)
display(qc_z.draw(output='mpl'))
sv_4 = Statevector.from_instruction(qc_z)
display(sv_4.draw('latex', prefix="|ψ⟩ = "))
```

## Rotation gates

```python
qc_rx = QuantumCircuit(1)
qc_rx.rx(pi/3, 0)
display(qc_rx.draw(output='mpl'))
sv_5 = Statevector.from_instruction(qc_rx)
display(sv_5.draw('latex', prefix="|ψ⟩ = "))
```

```python
qc_ry = QuantumCircuit(1)
qc_ry.ry(pi/4, 0)
display(qc_ry.draw(output='mpl'))
sv_6 = Statevector.from_instruction(qc_ry)
display(sv_6.draw('latex', prefix="|ψ⟩ = "))
```

```python
qc_rz = QuantumCircuit(1)
qc_rz.rz(pi/5, 0)
display(qc_rz.draw(output='mpl'))
sv_7 = Statevector.from_instruction(qc_rz)
display(sv_7.draw('latex', prefix="|ψ⟩ = "))
```

## Sequences of gates

```python
qc_seq = QuantumCircuit(1)
qc_seq.h(0)
qc_seq.s(0)
qc_seq.t(0)
display(qc_seq.draw(output='mpl'))
sv_8 = Statevector.from_instruction(qc_seq)
display(sv_8.draw('latex', prefix="|ψ⟩ = "))
```
