# One-Qubit Gates with Qiskit
This notebook builds a one‑qubit register, applies various gates, shows the statevector, and draws circuits.

```python
from math import pi
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from IPython.display import display, Markdown

def show_circuit_and_state(circ, title=None):
    if title:
        display(Markdown(f"### {title}"))
    try:
        display(circ.draw(output='mpl'))
    except Exception:
        print(circ.draw(output='text'))
    sv = Statevector.from_instruction(circ)
    print("Statevector:")
    print(sv.data)
```

## Basic gates

```python
qc_x = QuantumCircuit(1)
qc_x.x(0)
show_circuit_and_state(qc_x, "X Gate")
```

```python
qc_y = QuantumCircuit(1)
qc_y.y(0)
show_circuit_and_state(qc_y, "Y Gate")
```

```python
qc_z = QuantumCircuit(1)
qc_z.z(0)
show_circuit_and_state(qc_z, "Z Gate")
```

## Rotation gates

```python
qc_rx = QuantumCircuit(1)
qc_rx.rx(pi/3, 0)
show_circuit_and_state(qc_rx, "Rx(pi/3)")
```

```python
qc_ry = QuantumCircuit(1)
qc_ry.ry(pi/4, 0)
show_circuit_and_state(qc_ry, "Ry(pi/4)")
```

```python
qc_rz = QuantumCircuit(1)
qc_rz.rz(pi/5, 0)
show_circuit_and_state(qc_rz, "Rz(pi/5)")
```

## Sequences of gates

```python
qc_seq = QuantumCircuit(1)
qc_seq.h(0)
qc_seq.s(0)
qc_seq.t(0)
show_circuit_and_state(qc_seq, "H then S then T")
```
